```python
import numpy as np
rng = np.random.default_rng(43)
marks = rng.integers(0,101,size=(30,5))
print(marks)
```

```python
np.shape(marks)
```

```python
total = np.sum(marks,axis =1)
print(total)
```

```python
average = np.mean(marks, axis = 1)
print(average)
```

```python
# Highest Student

top_student = np.argmax(total)
print(" TOP STUDENT NO :",top_student)
highest = np.max(total)
print("HIGHEST MARKS IS :",highest)
```

```python
# Lowest Student
low_student = np.argmin(total)
print("LOWER STUDENT NO :",low_student)
lowest =np.min(total)
print("LOWEEST MARKS SI :", lowest)
```

```python
# Pass/Fail
passed= average >= 40
print(passed)
```

```python
# total passed
passed_student = np.sum(passed)
print(passed_student)
```

```python
# failed student
fail = average < 40
print(fail)
```

```python
# number of the student who is failed the exam
failed_students = np.where(average < 40)

print(failed_students)
```

```python
# Grade System
grade = np.where(average >= 80, 'A',
                 np.where(average >= 70, 'B',
                          np.where(average >= 60, 'C',
                                   np.where(average>= 50, 'D','F'
                                            ))))
print(grade)
```

```python
# average of the subject marks
subject_average = np.mean(marks, axis=0)

print(subject_average)
```
